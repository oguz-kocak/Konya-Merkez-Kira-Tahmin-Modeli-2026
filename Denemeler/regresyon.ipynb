```python
import pandas as pd
```

```python
df = pd.read_csv('statistically_cleaned_data.csv')
```

```python
df.district = df.district.astype('category')
df.neighborhood = df.neighborhood.astype('category')
```

### X ve y ozelliklerini belirleme

```python
X = df.drop(['price'],axis=1)
```

```python
y = df.price
```

### Train ve Test verilerini ayırma

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

### Kullanacagımız Olceklendiricilerden nesne üretme

```python
from sklearn.preprocessing import StandardScaler,OneHotEncoder
```

```python
numScaler = StandardScaler()
catScaler = OneHotEncoder(handle_unknown='ignore')
```

```python
numericalFeatures = [ 'room', 'living_room', 'area', 'age', 'floor']
categoricalFeatures = ['district', 'neighborhood']
```

### Train icin fit + transform

```python
numTrainScaled = numScaler.fit_transform(X_train[numericalFeatures])
```

```python
catTrainScaled = catScaler.fit_transform(X_train[categoricalFeatures])
```

```python
arrCatTrainScaled = catTrainScaled.toarray()
arrCatTrainScaled.astype(int)
```

```python
numTrainScaledDf = pd.DataFrame(numTrainScaled)
catTrainScaledDf = pd.DataFrame(arrCatTrainScaled)
```

```python
catTrainScaledDf = catTrainScaledDf.astype(int)
```

### scale edilmis degerleri tek bir data frame'de birlestirme

```python
TrainScaleEdilmisVeBirlestirilmisDf = pd.concat([numTrainScaledDf,catTrainScaledDf],axis=1)
```

```python
TrainScaleEdilmisVeBirlestirilmisDf
```

### Test icin sadece transform

```python
numTestScaled = numScaler.transform(X_test[numericalFeatures])
catTestScaled = catScaler.transform(X_test[categoricalFeatures])
```

```python
arrCatTestScaled = catTestScaled.toarray()
arrCatTestScaled.astype(int)
```

```python
numTestScaledDf = pd.DataFrame(numTestScaled)
catTestScaledDf = pd.DataFrame(arrCatTestScaled)
```

```python
catTestScaledDf = catTestScaledDf.astype(int)
```

```python
TestScaleEdilmisVeBirlestirimisDf = pd.concat([numTestScaledDf,catTestScaledDf],axis=1)
```
