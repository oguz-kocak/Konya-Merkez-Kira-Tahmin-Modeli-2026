```python
from sklearn.preprocessing import OneHotEncoder
```

```python
data3 =[
    ["Konya"],
    ["Ankara"],
    ["İstanbul"],
    ["Konya"]
]
```

```python
oneHotEncoder = OneHotEncoder(handle_unknown='ignore')
```

```python
try:
    categoricalDatas = oneHotEncoder.fit_transform(data3)
except Exception as e:
    print(e)
```

```python
categoricalDatas=categoricalDatas.toarray()
categoricalDatas = categoricalDatas.astype(int)
```

```python
numericalDatas = pd.DataFrame(numericalDatas)
categoricalDatas = pd.DataFrame(categoricalDatas)
```

```python
print(categoricalDatas)
```

```python
pd.concat([numericalDatas,categoricalDatas],axis=1)
```
