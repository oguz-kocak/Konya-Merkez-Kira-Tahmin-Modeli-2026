```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("hepsiemlak.csv")
```

```python
df['cityDistNeig'] = df['listingCard__media-link href'].apply(lambda x : x.split('/')[3])
```

```python
df['district'] = df['cityDistNeig'].apply(lambda x:x.split('-')[1])
df['neighborhood'] = df['cityDistNeig'].apply(lambda x:x.split('-')[2:-1])
```

```python
df['neighborhood'] = df['neighborhood'].apply(lambda x:"_".join(x))
```

```python
def dropColumn(df,colName):
    """ 
    sütünları siler

    parameters
    df = data frame 

    colName = column Name sütun adı
    """
    try:
        df.drop(colName,axis=1,inplace=True)
    except Exception as e:
        print(e)
```

```python
dropColumn(df,'listingCard__media-link href')
```

```python
silinecekler = ['listingCard__photo-index', 'listingCard__date',
       'listingCard__title', 'listingCard__spec-item', 'listingCard__spec-sep 4',
       'listingCard__spec-label 5',
       'listingCard__location', 'wp-btn-label', 'he-lazy-image src',
       'he-lazy-image src 2', 'listingCard__owner-firm', 'cityDistNeig',]
```

```python
dropColumn(df,silinecekler)
```

```python
df = df[df['listingCard__spec-item 2']!='Stüdyo']
```

```python
df['room'] = df['listingCard__spec-item 2'].apply(lambda x:x.split('+')[0])
df['living_room'] = df['listingCard__spec-item 2'].apply(lambda x:x.split('+')[1])
```

```python
dropColumn(df,'listingCard__spec-item 2')
```

```python
df['area'] = df['listingCard__spec-item 3'].apply(lambda x : x.split(' ')[0])
```

```python
dropColumn(df,'listingCard__spec-item 3')
```

```python
degisecekYaslar = {
    'Sıfır Bina' : '0 Yaşında'
}

df.replace(degisecekYaslar,inplace=True)
```

```python
df['age'] = df['listingCard__spec-item 4'].apply(lambda x : x.split(" ")[0])
```

```python
dropColumn(df,'listingCard__spec-item 4')
```

```python
df['listingCard__spec-item 5'].unique()
```

```python
degisecekKatlar = {
    'Giriş Katı' : '0. Kat',
    'Yüksek Giriş' : '0. Kat',
    'Ara Kat': '3 .Kat',
    'Zemin' : '0. Kat',
    'Kot 1': '-1. Kat',
    'Kot 2': '-2. Kat',
    'Kot 3': '-3. Kat',
    'Yarı Bodrum':'0. Kat',
    'Bahçe Katı' : '0. Kat',
    'Bodrum': '0. Kat',
    'Çatı Katı':'5. Kat',
    'En Üst Kat':'5. Kat',
    None : '0. Kat'
}
```

```python
df.replace(degisecekKatlar,inplace=True)
```

```python
df['floor'] = df['listingCard__spec-item 5'].apply(lambda x:x.split('.')[0])
```

```python
dropColumn(df,'listingCard__spec-item 5')
```

```python
df['price'] = df['listingCard__price']
```

```python
dropColumn(df,'listingCard__price')
```

```python
df['price'] = df.price * 1000
```
