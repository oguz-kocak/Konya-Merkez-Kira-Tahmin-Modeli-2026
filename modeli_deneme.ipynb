```python
import joblib
import pandas as pd

model = joblib.load("Konya-Merkez-Kira-Tahmin-Modeli.joblib")

yeni_ev = pd.DataFrame([{
    "district": "selcuklu",
    "neighborhood": "bosna_hersek",
    "room": 4,
    "living_room": 1,
    "area": 150,
    "age": 5,
    "floor": 3
}])

tahmin = model.predict(yeni_ev)
print(tahmin)
```
